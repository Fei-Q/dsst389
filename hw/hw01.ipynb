### Instructions

Draw a scatterplot with 4 points in which each point is a person. One person should be you. The others can be friends, family, or famous living people. Make the x-axis the person's age in years and the y-axis the name of the state or country where they were born (yes, the y-axis can be a categorical variable). Then, create the data to produce this plot in a hand-written table. Finally, write the Python code that would produce your plot from the data.

```python
data = {
    "name": ["Fei", "Alice", "Bill", "Elisa"],
    "age": [21, 17, 24, 20],
    "country": ["China", "Australia", "USA", "UK"]
}

df = pl.DataFrame(data)
df
```

```python
(
    df
    .pipe(ggplot, aes(x="age", y="country"))
    + geom_point()
    + geom_text(aes(label="name"), color="#08b8c4", size=10, nudge_y=0.2)
    + scale_x_continuous(breaks=True, limits=(15, 25))
    + theme(figure_size=(6, 4))
)
```
